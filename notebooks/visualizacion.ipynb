# Visualización del dataset Iris

Este cuaderno replica las visualizaciones de la aplicación Streamlit: una gráfica 2D de largo del pétalo frente a ancho del pétalo y una gráfica 3D con tres características.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from src.data_loader import get_iris_data
```

## Carga del dataset completo

El loader existente devuelve una partición reproducible. Se concatenan ambas partes para que las visualizaciones incluyan las 150 observaciones.

```python
X_train, X_test, y_train, y_test = get_iris_data()
features = pd.concat([X_train, X_test], ignore_index=True)
targets = pd.Series(
    pd.concat([pd.Series(y_train), pd.Series(y_test)], ignore_index=True),
    name="class",
)
features.head()
```

## Gráfica 2D: Largo del Pétalo vs Ancho del Pétalo

```python
plot_data = features[["petal length", "petal width"]].copy()
plot_data["class"] = targets.to_numpy()

figure, axis = plt.subplots(figsize=(9, 5))
sns.scatterplot(
    data=plot_data,
    x="petal length",
    y="petal width",
    hue="class",
    palette="viridis",
    ax=axis,
)
axis.set_title("Largo del Pétalo vs Ancho del Pétalo")
axis.set_xlabel("Largo del Pétalo (LP)")
axis.set_ylabel("Ancho del Pétalo (AP)")
axis.grid(alpha=0.25)
figure.tight_layout()
plt.show()
```

## Gráfica 3D

Se utilizan largo del sépalo, ancho del sépalo y largo del pétalo, coloreando cada observación según su clase.

```python
features_3d = ["sepal length", "sepal width", "petal length"]
+    figure = plt.figure(figsize=(9, 6))
axis = figure.add_subplot(111, projection="3d")
scatter = axis.scatter(
    features[features_3d[0]],
    features[features_3d[1]],
    features[features_3d[2]],
    c=pd.Categorical(targets).codes,
    cmap="viridis",
    alpha=0.8,
)
axis.set_title("Visualización 3D del Dataset Iris")
axis.set_xlabel("Largo del Sépalo")
axis.set_ylabel("Ancho del Sépalo")
axis.set_zlabel("Largo del Pétalo")
figure.colorbar(scatter, ax=axis, pad=0.1, label="Clase")
figure.tight_layout()
plt.show()
```
